1. Daily Weather Logger Simulation (CSV + Delay Scraping)

```python
import requests
from bs4 import BeautifulSoup
import csv
import time
from datetime import datetime

BASE_URL = "https://www.timeanddate.com/weather/philippines/"

city = input("Enter city name: ").strip().lower().replace(" ", "-")
days_input = input("Number of days (default 3): ")

try:
    num_days = int(days_input)
    if num_days < 1 or num_days > 7:
        num_days = 3
except:
    num_days = 3

url = BASE_URL + city
headers = {"User-Agent": "Mozilla/5.0"}

file_name = "weather_data.csv"

file_exists = False
try:
    with open(file_name, "r"):
        file_exists = True
except:
    file_exists = False

for i in range(num_days):
    print(f"Scraping run {i+1}...")

    try:
        response = requests.get(url, headers=headers, timeout=10)
        soup = BeautifulSoup(response.text, "lxml")

        # --- BASIC EXTRACTION (structure may vary depending on page) ---
        temp = soup.find("div", class_="h2").text.strip() if soup.find("div", class_="h2") else "N/A"
        condition = soup.find("p").text.strip() if soup.find("p") else "N/A"

        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        row = [
            timestamp, city, "today",
            temp, "N/A", "N/A",
            condition, "N/A", "N/A",
            "N/A", "N/A", "N/A",
            "N/A"
        ]

        with open(file_name, "a", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)

            if not file_exists:
                writer.writerow([
                    "timestamp", "city", "forecast_date",
                    "temperature", "temperature_min", "temperature_max",
                    "condition", "wind", "humidity",
                    "pressure", "visibility", "dew_point",
                    "feels_like"
                ])
                file_exists = True

            writer.writerow(row)

        print("Saved one weather record.")

    except Exception as e:
        print("Error:", e)

    time.sleep(10)

print("Done logging weather data.")
```
