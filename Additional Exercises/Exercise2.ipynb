2. JSON Weather Exporter

```python
import requests
from bs4 import BeautifulSoup
import json
from datetime import datetime

BASE_URL = "https://www.timeanddate.com/weather/philippines/"

city = input("Enter city name: ").strip().lower().replace(" ", "-")
days_input = input("Forecast days (default 3): ")

try:
    num_days = int(days_input)
    if num_days < 1 or num_days > 7:
        num_days = 3
except:
    num_days = 3

url = BASE_URL + city
headers = {"User-Agent": "Mozilla/5.0"}

data_list = []

try:
    response = requests.get(url, headers=headers, timeout=10)
    soup = BeautifulSoup(response.text, "lxml")

    temp = soup.find("div", class_="h2").text.strip() if soup.find("div", class_="h2") else "N/A"
    condition = soup.find("p").text.strip() if soup.find("p") else "N/A"

    record = {
        "timestamp": str(datetime.now()),
        "city": city,
        "forecast_date": "today",
        "temperature": temp,
        "temperature_min": "N/A",
        "temperature_max": "N/A",
        "condition": condition,
        "wind": "N/A",
        "humidity": "N/A",
        "pressure": "N/A",
        "visibility": "N/A",
        "dew_point": "N/A",
        "feels_like": "N/A"
    }

    data_list.append(record)

    file_name = "weather_data.json"

    try:
        with open(file_name, "r") as f:
            existing_data = json.load(f)
    except:
        existing_data = []

    existing_data.extend(data_list)

    with open(file_name, "w", encoding="utf-8") as f:
        json.dump(existing_data, f, indent=4)

    print("Saved to weather_data.json")

except Exception as e:
    print("Error:", e)
```
